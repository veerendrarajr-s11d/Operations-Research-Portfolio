## The Continuous Textile Factory (Linear Programming)

**The Business Problem:**
We are managing a textile factory producing Suiting, Shirting, and Woolen fabrics. 
Our goal is to maximize total profit, but we are strictly limited by the available hours on our Weaving, Processing, and Finishing machines.

* **Profits:** Suiting (Rs. 2/m), Shirting (Rs. 4/m), Woolen (Rs. 3/m)
* **Weaving Limit:** 3600 hours available
* **Processing Limit:** 2400 hours available
* **Finishing Limit:** 4800 hours available

**The Mathematical Objective:** 
Calculate the exact production volume for each fabric to guarantee the absolute maximum profit without exceeding machine capacity.

```python
from scipy.optimize import linprog
import numpy as np

print("--- CONTINUOUS FACTORY LINEAR PROGRAMMING ---")

c_continuous = [-2, -4, -3]  

A_ub = [
    [3, 4, 3], 
    [2, 1, 3], 
    [1, 3, 3]  
]
b_ub = [3600, 2400, 4800] 

res_cont = linprog(c_continuous, A_ub=A_ub, b_ub=b_ub, method='highs')

print(f"Optimal Production (Decimals): {np.round(res_cont.x, 2)}")
print(f"Max Profit: Rs. {abs(res_cont.fun)}")
```

## Stage 5: Integer Programming (AEROSPACE)

**The Business Problem:**
We manage an aerospace factory building Light, Midsize, and Heavy jets. Our production is constrained by our inventory of Aluminum, Titanium, and Carbon Fiber. Unlike continuous textiles, we cannot sell 3.5 airplanes. Optimization must handle situations where fractional results are impossible.

**The Mathematical Objective:** 
Apply Branch and Bound algorithms by introducing `integrality` parameters to the linear programming model. This forces the decision variables to be strictly whole numbers (integers), forcing the solver to calculate the true, physical maximum profit.

```python

shadow_prices = res_cont.ineqlin.marginals
print(f"Shadow Prices (Marginal Value of Hours): {np.abs(shadow_prices)}")


print("--- INTEGER PROGRAMMING (AEROSPACE) ---")
# Objective: Maximize Profit (Light Jets, Midsize Jets, Heavy Jets)
c_integer = [-50, -40, -60]

# Constraints: Aluminum, Titanium, Carbon Fiber limits
A_ub_int = [
    [10, 8, 12],
    [5,  4,  8],
    [2,  3,  4]
]
b_ub_int = [100, 50, 30]

# Enforce strict whole-number rule for airplanes (1 = Integer, 0 = Continuous)
integrality = [1, 1, 1] 

res_int = linprog(c_integer, A_ub=A_ub_int, b_ub=b_ub_int, integrality=integrality, method='highs')

print(f"Optimal Production (Whole Jets Only): {res_int.x}")
print(f"Max Profit: Rs. {abs(res_int.fun)} Million")
```

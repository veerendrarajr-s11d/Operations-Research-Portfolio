## Stage 6: Competitive Strategy & Game Theory

**The Business Problem:**
We are the Pricing Director for Swiggy, entering a new market where Zomato is directly competing against us. Both companies have two primary moves: Run Huge Ads, or Offer Steep Discounts. 

Because our payoff (new customers won) depends entirely on what Zomato does simultaneously, there is no single "winning" move. If we become predictable, we lose.

**The Mathematical Objective:** 
Apply the Minimax Theorem using Linear Programming. We will calculate the perfect probabilistic "Mixed Strategy" (e.g., executing Ads X% of the time and Discounts Y% of the time). This mathematically guarantees Swiggy a maximum baseline market share (Value of the Game, V) regardless of the counter-strategies Zomato deploys.

```python
from scipy.optimize import linprog

print("--- GAME THEORY (MINIMAX STRATEGY) ---")
c_game = [-1, 0, 0] #[cite: 7]
 
A_ub_game = [
    [1, -4, -1],  # Opponent plays Strategy 1
    [1, -2, -5]   # Opponent plays Strategy 2
]
b_ub_game = [0, 0]
A_eq_game = [[0, 1, 1]]
b_eq_game = [1]

bounds_game = [(None, None), (0, 1), (0, 1)]

res_game = linprog(c_game, A_ub=A_ub_game, b_ub=b_ub_game, A_eq=A_eq_game, b_eq=b_eq_game, bounds=bounds_game, method='highs')

print(f"Guaranteed Minimum Value (V): {round(res_game.x[0], 2)}")
print(f"Probability of Strategy 1: {round(res_game.x[1] * 100, 1)}%")
print(f"Probability of Strategy 2: {round(res_game.x[2] * 100, 1)}%")
```
